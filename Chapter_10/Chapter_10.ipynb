# Chapter 10: Robustness

-   *Productionising* code is the process of taking simple programs or
    ad-hoc scripts and converting them into long-term maintainable code
-   This involves making programs that are capable of handling
    unexpected circumstances e.g.
    -   Unexpected input
    -   Missing data
    -   Unexpected control flow
    -   Use of a function or module outside of it’s original context
-   Making programs safe is as important as making them work
