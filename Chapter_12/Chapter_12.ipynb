# Chapter 12: Data Structures and Algorithms

-   For non-trivial data manipulation implementing the correct algorithm
    is important to maintain performance
-   Many standard data structures and algorithms are prepackaged by the
    python standard library
-   More complex pre-packaged data structures and types also exist for
    handling specialised data, including
    1.  Date and time manipulations
    2.  Handling monetary or other high-precision decimal values
    3.  Saving and restoring program state, and more
-   Where possible prefer using library-implementations over rolling
    your own
